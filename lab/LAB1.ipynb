```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats
```

```python
## QUESTION 1
```

```python
# The cumulative distribution function gives the probability of a value (X) being less
# a specified value (x). It does this by integrating the normal distribution curve, taking 
# the value of the are under the curve as the probability. So we can use this to find the 
# integral of the normal distribution.

# for standard normal distribution set mean=0 and standard deviation=1
p = stats.norm.cdf(x=-3.61, loc=0, scale=1)
print(p)

# trials of x=1.4, 0.25, -3.61 match z-table
```

```python
# Now, we input a probability to get a z-score.

s1 = stats.norm.ppf(0.95)
print(s1)

# matches z-table since a z-score of 1.64 gives a probability of 0.94950 and one of 1.65 gives 0.95053
```

```python
s2 = stats.norm.ppf(0.27)
print(s2)

# matches z-table since a z-score of -0.61 gives probability of 0.27093
```

```python
# The negative z-score signifies the x position on a standard normal distribution (centered at x=0).
# This means the probability associated with it will be less than 50%.
```

```python
# set plot size
plt.rcParams["figure.figsize"] = (15,10)
```

```python
d = stats.rayleigh.rvs(loc = 0, scale = 1, size = 1000000)
```

```python
fig, ax = plt.subplots(1, 1)

# plot realization histogram of 100k Rayleigh d sample
ax.hist(d,50, density=True)
plt.tick_params(labelsize = 24)

# adjust axis limits to display plot nicely
plt.xlim([-0.05,5])

# plot analytical pdf of Rayleigh distribution
x = np.linspace(-.05,5,1000)
ax.plot(x,stats.rayleigh.pdf(x,loc = 0, scale = 1),linewidth = 8,alpha = 0.7)

# show plot
plt.show()
```

```python
# Now plotting both the analytical pdf and realization histogram on a logy scale

fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([-.5,5.5])
x = np.linspace(-0.5,5.5,1000)
ax.plot(x,stats.rayleigh.pdf(x,loc = 0, scale = 1),linewidth = 8,alpha = 0.7)
plt.show()
```
