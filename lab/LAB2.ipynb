```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats
```

```python
# Problem 1(a)
```

```python
x = 2 #k
y = 5 #mu or lambda

xaxis = np.arange(0,10)

prob = stats.poisson.pmf(xaxis,x)
print(sum(prob))

fig, ax = plt.subplots(1, 1)


ax.step (xaxis, prob, where='mid')
plt.xlabel("number of background in one day")
plt.ylabel("probability")
plt.show()
```

```python
twodays= np.convolve(prob,prob, mode = 'full')
print(sum(twodays))
threedays = np.convolve(twodays, prob, mode = 'full')
print(sum(threedays))
fig, ax = plt.subplots(1, 1)
ax.step (np.arange(0,len(twodays),1), twodays, where='mid')
ax.step (np.arange(0,len(threedays),1), threedays, where='mid')
plt.show()
```

```python
# Problem 1(b)
```

```python
twodays= np.convolve(prob,prob, mode = 'full')
threedays = np.convolve(twodays, prob, mode = 'full')
fourdays = np.convolve(threedays, prob, mode = 'full')
fivedays = np.convolve(fourdays, prob, mode = 'full')

print(sum(fourdays))
print(sum(fivedays))
xval_twodays_sum = np.arange(0,len(twodays),1)



fig, ax = plt.subplots(1, 1)
ax.step (xval_twodays_sum, twodays, where='mid')
ax.step (np.arange(0,len(threedays),1), threedays, where='mid')
ax.step (np.arange(0,len(fourdays),1), fourdays, where='mid')
ax.step (np.arange(0,len(fivedays),1), fivedays, where='mid')
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.step (np.arange(0,len(fivedays),1), fivedays, where='mid')
ax.set_yscale('log')
plt.show()
```

Since the distribution in a semi-logy graph is not a perfect upside down quadratic, it is still not a perfect gaussian and is thus still closer to a Poisson distribution.
