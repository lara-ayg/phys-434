```python
# I am picking a Rayleigh distribution and an observing interval of 1 day.
```

```python
xaxisr = np.linspace(0, stats.rayleigh.ppf(0.999), 2000)

# Calculate the probability density at every x value
probr = stats.rayleigh.pdf(xaxisr)

fig, ax = plt.subplots(1, 1)


ax.step (xaxisr, probr, where='mid')
plt.xlabel("number of background in one day")
plt.ylabel("probability density")
plt.show()
```

```python
# spacing between the original x values
dx = xaxisr[1] - xaxisr[0]

# convolve the daily Rayleigh density
twodaysr = np.convolve(probr, probr, mode="full") * dx
threedaysr = np.convolve(twodaysr, probr, mode="full") * dx
fourdaysr = np.convolve(threedaysr, probr, mode="full") * dx
fivedaysr = np.convolve(fourdaysr, probr, mode="full") * dx
sixdaysr = np.convolve(fivedaysr, probr, mode="full") * dx
sevendaysr = np.convolve(sixdaysr, probr, mode="full") * dx
eightdaysr = np.convolve(sevendaysr, probr, mode="full") * dx
ninedaysr = np.convolve(eightdaysr, probr, mode="full") * dx
tendaysr = np.convolve(eightdaysr, probr, mode="full") * dx

# x-axis for the sum of each number of observing intervals
x2 = np.arange(len(twodaysr)) * dx + 2 * xaxisr[0]
x3 = np.arange(len(threedaysr)) * dx + 3 * xaxisr[0]
x4 = np.arange(len(fourdaysr)) * dx + 4 * xaxisr[0]
x5 = np.arange(len(fivedaysr)) * dx + 5 * xaxisr[0]
x6 = np.arange(len(sixdaysr)) * dx + 6 * xaxisr[0]
x7 = np.arange(len(sevendaysr)) * dx + 7 * xaxisr[0]
x8 = np.arange(len(eightdaysr)) * dx + 8 * xaxisr[0]
x9 = np.arange(len(ninedaysr)) * dx + 9 * xaxisr[0]
x10 = np.arange(len(tendaysr)) * dx + 10 * xaxisr[0]

fig, ax = plt.subplots()

# plot the average background per day
ax.plot(x2 / 2, 2 * twodaysr)
ax.plot(x3 / 3, 3 * threedaysr)
ax.plot(x4 / 4, 4 * fourdaysr)
ax.plot(x5 / 5, 5 * fivedaysr)
ax.plot(x6 / 6, 6 * sixdaysr)
ax.plot(x7 / 7, 7 * sevendaysr)
ax.plot(x8 / 8, 8 * eightdaysr)
ax.plot(x9 / 9, 9 * ninedaysr)
ax.plot(x10 / 10, 10 * tendaysr)
plt.xlabel("average background per day")
plt.ylabel("probability density")
ax.legend(["2 days", "3 days", "4 days", "5 days","6 days", "7 days","8 days", "9 days","10 days"])
plt.show()
```

```python
# for loop to make it easier to see more convolutions
distributions = [probr]
x_values = [xaxisr]

for n in range(2, 51):
    # Convolve with the original one-day distribution
    next_dist = np.convolve(distributions[-1], probr, mode="full") * dx
    distributions.append(next_dist)

    # x-axis for the sum of n observing intervals
    x_sum = np.arange(len(next_dist)) * dx + n * xaxis[0]
    x_values.append(x_sum)

# plot average background per day
fig, ax = plt.subplots(figsize=(10, 6))

for n in range(2, 51):
    ax.plot(
        x_values[n-1] / n,
        n * distributions[n-1],
        label=f"{n} days"
    )

ax.set_xlabel("Average background per day")
ax.set_ylabel("Probability density")
plt.show()
```

```python
# Over more and more intervals (days), the distribution approaches a more normal distribution.
```

```python
# Background distribution
mu = 0
sigma = 2.0

# Global false-alarm probability for a 5-sigma detection
p_global_target = stats.norm.sf(5)

# Trials factors spanning several orders of magnitude
trials = [1, 10, 100, 1000, 10000, 100000, 1000000]

thresholds = []
z_scores = []

for N in trials:
    # Find the required single-pixel p-value
    p_pixel = -np.expm1(np.log1p(-p_global_target) / N)
    # Convert p-value to a z-score
    z = stats.norm.isf(p_pixel)
    # Convert z-score to required brightness
    Y = mu + sigma * z
    z_scores.append(z)
    thresholds.append(Y)

print("Trials factor | Z threshold | Brightness threshold")
for N, z, Y in zip(trials, z_scores, thresholds):
    print(f"{N:13,d} | {z:10.2f} | {Y:19.2f}")

```

```python
# The effect isn't that great because even increasing the trials factor from
# 1 to 1000000 only increases the brightness threshold from 10 to 14.3
# this makes sense because Gaussians decrease really quick so only a small
# increase in threshold is needed for a lot of extra trials.
```
