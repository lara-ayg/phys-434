```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
```

```python
# generating the data points for the histogram, centered at x=3 with 1000000 data points
d = stats.norm.rvs(loc = 3.0, scale = .01, size = 1000000)
```

```python
fig, ax = plt.subplots(1, 1)
#histogram of d from above
ax.hist(d,50, density=True)
#tick marks
plt.tick_params(labelsize = 24)
# x-axis limits
plt.xlim([1.95,4.05])

# plotting probability density function (orange line)
x = np.linspace(1.95,4.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 3., scale = 0.01),linewidth = 4,alpha = 0.7)
plt.show()
```

```python
ax = plt.hist(d,50, density=True)
#same histogram just on log scale
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([1.95,4.05])
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.yscale('log')
#log y-scale checks whether distribution is gaussian
plt.tick_params(labelsize = 24)
plt.xlim([1.95,4.05])
x = np.linspace(1.95,4.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 3., scale = 0.01),linewidth = 1,alpha = 0.7)
plt.show()
```
